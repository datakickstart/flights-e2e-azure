```sql
CREATE OR REFRESH STREAMING TABLE main.dustinvannoy_dev.airlines_st_silver
TBLPROPERTIES (
  'delta.enableDeletionVectors' = true,
  'delta.enableChangeDataFeed' = true
) AS
SELECT
  concat( cast(Year as string),'-',
          lpad(cast(Month as string),2,'0'),'-',
          lpad(cast(DayOfMonth as string),2,'0')
    ) event_date,
  try_to_timestamp(concat(event_date, ' ', 
    concat(left(CRSDepTime, 2), ':', right(CRSDepTime, 2)))
    ) as `timestamp`,
  Year,
  Month,
  DayofMonth,
  DayOfWeek,
  lpad(DepTime,4, '0') as DepTime,
  CRSDepTime,
  ArrTime,
  replace(CRSArrTime, 'NA') as CRSArrTime,
  UniqueCarrier,
  FlightNum,
  TailNum,
  ActualElapsedTime,
  CRSElapsedTime,
  AirTime,
  ArrDelay,
  DepDelay,
  Origin,
  Dest,
  Distance,
  TaxiIn,
  TaxiOut,
  Cancelled,
  CancellationCode,
  Diverted,
  CarrierDelay,
  WeatherDelay,
  NASDelay,
  SecurityDelay,
  LateAircraftDelay,
  IsArrDelayed,
  IsDepDelayed,
  created_timestamp_ltz,
  current_timestamp() as updated_timestamp
FROM
  STREAM main.dustinvannoy_dev.airlines_st_bronze
```

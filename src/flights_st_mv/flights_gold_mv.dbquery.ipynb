```sql
CREATE OR REPLACE MATERIALIZED VIEW main.dustinvannoy_dev.flights_carrier_num_agg_mv_gold AS
SELECT UniqueCarrier, FlightNum, 1 as flight_count
FROM main.dustinvannoy_dev.airlines_st_silver
GROUP BY ALL;

CREATE OR REPLACE MATERIALIZED VIEW main.dustinvannoy_dev.flights_totals_daily_mv_gold AS
SELECT event_date, COUNT(DISTINCT UniqueCarrier) carriers, COUNT(DISTINCT FlightNum) flights
FROM main.dustinvannoy_dev.airlines_st_silver
GROUP BY ALL;

REFRESH MATERIALIZED VIEW main.dustinvannoy_dev.flights_carrier_num_agg_mv_gold;

REFRESH MATERIALIZED VIEW main.dustinvannoy_dev.flights_totals_daily_mv_gold;


```
